# Restaurant Sales & Tipping Analytics Dashboard
### Data Visualisation Activity

**Objective:** Build an interactive dashboard to explore restaurant bills, tips, dining periods, and customer party sizes using Python and Plotly.

**Tools:** Python, Pandas, Plotly, Google Colab

**Dataset:** Plotly's built-in Tips dataset (244 records).

This notebook completes the activity by:
1. Loading and inspecting the dataset.
2. Performing basic data analysis.
3. Creating five different chart types.
4. Adding interactive hover, zoom, and pan features.
5. Exporting all charts into one standalone HTML dashboard.

> Run the cells from top to bottom in Google Colab. The HTML export uses Plotly's CDN, so an internet connection is required to view the charts.

## Activity 1: Install and import required libraries

```python
!pip -q install plotly pandas numpy

import pandas as pd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go
from IPython.display import display, HTML
from google.colab import files

print("Required libraries imported successfully.")
```

## Activity 2: Load the dataset

```python
# Load Plotly's built-in restaurant tips dataset
df = px.data.tips()

print("Dataset loaded successfully.")
print("Number of records:", len(df))
print("Number of columns:", len(df.columns))

display(df.head())
```

## Activity 3: Explore and understand the data

```python
# Inspect column names, data types, and missing values
print("Dataset information:")
df.info()

print("\nMissing values in each column:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nNumerical summary:")
display(df.describe())
```

## Activity 4: Prepare the data and calculate summary statistics

```python
# Check for duplicate records
print("Duplicate rows:", df.duplicated().sum())

# Create a tip percentage column for additional analysis
df["tip_percentage"] = (df["tip"] / df["total_bill"]) * 100

# Calculate summary values
average_bill = df["total_bill"].mean()
average_tip = df["tip"].mean()
average_party_size = df["size"].mean()
average_tip_percentage = df["tip_percentage"].mean()

summary = pd.DataFrame({
    "Metric": ["Average bill", "Average tip", "Average party size", "Average tip percentage"],
    "Value": [
        round(average_bill, 2),
        round(average_tip, 2),
        round(average_party_size, 2),
        round(average_tip_percentage, 2)
    ]
})

display(summary)
```

## Activity 5: Create five different visualisations

A consistent dark theme and readable titles are used throughout. Plotly charts support hover tooltips, zooming, and panning.

### Chart 1: Bar chart - Average tips by day

```python
daily_tips = df.groupby("day", as_index=False)["tip"].mean()

fig1 = px.bar(
    daily_tips,
    x="day",
    y="tip",
    color="tip",
    color_continuous_scale="Viridis",
    title="Average Tips by Day",
    labels={"day": "Day of the Week", "tip": "Average Tip"}
)
fig1.update_layout(template="plotly_dark", title_x=0.5, height=430)
fig1.show()
```

### Chart 2: Scatter plot - Total bill versus tip

```python
fig2 = px.scatter(
    df,
    x="total_bill",
    y="tip",
    color="time",
    size="size",
    hover_data=["day", "sex", "smoker"],
    title="Relationship Between Total Bill and Tip",
    labels={"total_bill": "Total Bill", "tip": "Tip", "time": "Dining Period"}
)
fig2.update_layout(template="plotly_dark", title_x=0.5, height=430)
fig2.show()
```

### Chart 3: Histogram - Distribution of restaurant bills

```python
fig3 = px.histogram(
    df,
    x="total_bill",
    nbins=25,
    color_discrete_sequence=["#00CC96"],
    title="Distribution of Restaurant Bills",
    labels={"total_bill": "Total Bill", "count": "Number of Customers"}
)
fig3.update_layout(template="plotly_dark", title_x=0.5, height=430)
fig3.show()
```

### Chart 4: Box plot - Tips during lunch and dinner

```python
fig4 = px.box(
    df,
    x="time",
    y="tip",
    color="time",
    points="outliers",
    title="Tip Distribution: Lunch vs Dinner",
    labels={"time": "Dining Period", "tip": "Tip"}
)
fig4.update_layout(template="plotly_dark", title_x=0.5, height=430)
fig4.show()
```

### Chart 5: Heatmap - Correlation between numerical variables

```python
corr = df[["total_bill", "tip", "size", "tip_percentage"]].corr()

fig5 = go.Figure(data=go.Heatmap(
    z=corr.values,
    x=corr.columns,
    y=corr.columns,
    colorscale="Viridis",
    zmin=-1,
    zmax=1,
    text=np.round(corr.values, 2),
    texttemplate="%{text}",
    hovertemplate="%{x} vs %{y}<br>Correlation: %{z:.2f}<extra></extra>"
))

fig5.update_layout(
    title="Correlation Between Numerical Variables",
    title_x=0.5,
    template="plotly_dark",
    height=480
)
fig5.show()
```

## Activity 6: Create and export the interactive dashboard

```python

# INTERACTIVE RESTAURANT ANALYTICS DASHBOARD

import plotly.express as px
import plotly.graph_objects as go
import plotly.io as pio
import pandas as pd
import ipywidgets as widgets

from IPython.display import display, HTML, clear_output
from google.colab import output as colab_output

# Enable Google Colab widgets
colab_output.enable_custom_widget_manager()

# Set Plotly renderer
pio.renderers.default = "colab"

# Prepare dataset
dashboard_df = px.data.tips()

dashboard_df["tip_percentage"] = (
    dashboard_df["tip"] / dashboard_df["total_bill"]
) * 100

# Dashboard heading
display(HTML("""
<div style="background:#111827;padding:25px;border-radius:12px;
color:white;text-align:center;">
<h1 style="color:#00CC96;">Restaurant Analytics Dashboard</h1>
<p>Interactive Sales and Customer Tipping Analysis</p>
<p>Explore restaurant spending and tipping patterns</p>
</div>
"""))

# Create filters
day_filter = widgets.Dropdown(
    options=["All"] + list(dashboard_df["day"].unique()),
    value="All",
    description="Day:"
)

time_filter = widgets.Dropdown(
    options=["All", "Lunch", "Dinner"],
    value="All",
    description="Time:"
)

gender_filter = widgets.Dropdown(
    options=["All"] + list(dashboard_df["sex"].unique()),
    value="All",
    description="Gender:"
)

reset_button = widgets.Button(
    description="Reset Filters",
    button_style="warning",
    icon="refresh"
)

chart_output = widgets.Output()

# Filter dataset
def get_filtered_data():

    data = dashboard_df.copy()

    if day_filter.value != "All":
        data = data[data["day"] == day_filter.value]

    if time_filter.value != "All":
        data = data[data["time"] == time_filter.value]

    if gender_filter.value != "All":
        data = data[data["sex"] == gender_filter.value]

    return data


# Update dashboard
def update_dashboard(change=None):

    data = get_filtered_data()

    with chart_output:

        clear_output(wait=True)

        if data.empty:
            display(HTML(
                "<h3>No records found for selected filters.</h3>"
            ))
            return

        # KPI calculations
        revenue = data["total_bill"].sum()
        avg_bill = data["total_bill"].mean()
        avg_tip = data["tip"].mean()
        transactions = len(data)

        # KPI cards
        display(HTML(f"""
        <div style="
            display:grid;
            grid-template-columns:repeat(4,minmax(0,1fr));
            gap:12px;
            margin:15px 0;
        ">

        <div style="background:#1F2937;padding:18px;
        border-radius:10px;color:white;text-align:center;">
            <h4>Total Revenue</h4>
            <h2>${revenue:,.2f}</h2>
        </div>

        <div style="background:#1F2937;padding:18px;
        border-radius:10px;color:white;text-align:center;">
            <h4>Average Bill</h4>
            <h2>${avg_bill:.2f}</h2>
        </div>

        <div style="background:#1F2937;padding:18px;
        border-radius:10px;color:white;text-align:center;">
            <h4>Average Tip</h4>
            <h2>${avg_tip:.2f}</h2>
        </div>

        <div style="background:#1F2937;padding:18px;
        border-radius:10px;color:white;text-align:center;">
            <h4>Transactions</h4>
            <h2>{transactions}</h2>
        </div>

        </div>
        """))

        # Common chart styling
        layout = {
            "template": "plotly_dark",
            "height": 420,
            "margin": dict(l=45, r=25, t=65, b=45),
            "font": dict(family="Arial", size=13)
        }

        # 1. BAR CHART
        bar_data = data.groupby(
            "day", as_index=False
        )["tip"].mean()

        fig1 = px.bar(
            bar_data,
            x="day",
            y="tip",
            color="tip",
            color_continuous_scale="Viridis",
            title="Average Tips by Day",
            hover_data={"tip":":.2f"}
        )

        fig1.update_layout(**layout)
        display(fig1)

        # 2. SCATTER PLOT
        fig2 = px.scatter(
            data,
            x="total_bill",
            y="tip",
            color="time",
            size="size",
            hover_data=["day", "sex", "smoker"],
            title="Total Bill vs Tip"
        )

        fig2.update_layout(**layout)
        display(fig2)

        # 3. HISTOGRAM
        fig3 = px.histogram(
            data,
            x="total_bill",
            nbins=25,
            color_discrete_sequence=["#00CC96"],
            title="Distribution of Restaurant Bills"
        )

        fig3.update_layout(**layout)
        display(fig3)

        # 4. BOX PLOT
        fig4 = px.box(
            data,
            x="time",
            y="tip",
            color="time",
            points="outliers",
            title="Lunch vs Dinner Tip Distribution"
        )

        fig4.update_layout(**layout)
        display(fig4)

        # 5. HEATMAP
        corr = data[
            ["total_bill", "tip", "size", "tip_percentage"]
        ].corr()

        fig5 = go.Figure(
            data=go.Heatmap(
                z=corr.values,
                x=corr.columns,
                y=corr.columns,
                colorscale="Viridis",
                text=corr.round(2).values,
                texttemplate="%{text}",
                hovertemplate=
                "%{x} vs %{y}<br>Correlation: %{z:.2f}<extra></extra>"
            )
        )

        fig5.update_layout(
            title="Correlation Heatmap",
            **layout
        )

        display(fig5)


# Reset filters
def reset_filters(button):

    day_filter.value = "All"
    time_filter.value = "All"
    gender_filter.value = "All"


reset_button.on_click(reset_filters)

# Connect filters to dashboard
day_filter.observe(update_dashboard, names="value")
time_filter.observe(update_dashboard, names="value")
gender_filter.observe(update_dashboard, names="value")

# Dashboard controls
controls = widgets.HBox([
    day_filter,
    time_filter,
    gender_filter,
    reset_button
])

# Display dashboard
display(controls)
display(chart_output)

# Generate initial dashboard
update_dashboard()

print("Interactive dashboard loaded successfully!")
```

```python

# EXPORT INTERACTIVE DASHBOARD AS HTML

import plotly.io as pio
from IPython.display import display, HTML
from google.colab import files

# Collect all five charts
charts = [fig1, fig2, fig3, fig4, fig5]

# Create HTML sections
chart_sections = []

for i, fig in enumerate(charts):

    chart_html = pio.to_html(
        fig,
        full_html=False,
        include_plotlyjs="cdn" if i == 0 else False,
        config={
            "responsive": True,
            "displayModeBar": True
        }
    )

    chart_sections.append(
        f'<div class="chart">{chart_html}</div>'
    )

# Build HTML dashboard
html_content = f'''
<!DOCTYPE html>
<html>
<head>
<title>Restaurant Analytics Dashboard</title>

<style>
body {{
    background: #111827;
    color: white;
    font-family: Arial, sans-serif;
    padding: 20px;
}}

header {{
    text-align: center;
    color: #00CC96;
    padding: 20px;
}}

.dashboard {{
    display: grid;
    grid-template-columns: repeat(2, minmax(0, 1fr));
    gap: 20px;
}}

.chart {{
    background: #1F2937;
    padding: 10px;
    border-radius: 12px;
    min-width: 0;
}}

.chart:last-child {{
    grid-column: 1 / -1;
}}

@media(max-width: 800px) {{
    .dashboard {{
        grid-template-columns: 1fr;
    }}
    .chart:last-child {{
        grid-column: auto;
    }}
}}
</style>
</head>

<body>

<header>
<h1>Restaurant Sales & Tipping Analytics</h1>
<p>Interactive Data Visualisation Dashboard</p>
</header>

<div class="dashboard">
{''.join(chart_sections)}
</div>

</body>
</html>
'''

# Save HTML file
filename = "Restaurant_Analytics_Dashboard.html"

with open(filename, "w", encoding="utf-8") as f:
    f.write(html_content)

print("Dashboard exported successfully!")

# Download dashboard
files.download(filename)
```
